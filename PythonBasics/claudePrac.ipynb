new_list = [expression for item in iterable]
new_list = [expression for item in iterable if condition]

if condition: 
-This acts as a gatekeeper. The expression is only evaluated and added to the list if the condition evaluates to True.

This is where many beginners get tripped up. If you want to change the output based on a condition (rather than filtering items out entirely), the if-else block moves to the beginning of the comprehension, right inside the expression.

The Formula:
new_list = [expression_if_true if condition else expression_if_false for item in iterable]

````python
def clip(samples, limit):
    for i in range(samples.size()):
        if samples[i] > limit:
            samples[i] = limit
        elif samples[i] < -limit and samples[i] != 0 :
            samples[i] = -limit
    return samples
````

results=[] as a default argument. 
Python evaluates default arguments once, at function definition time — not on every call.
 
That list is created once and reused across every call that doesn't pass its own results.
Proof from the run: calling process([1,2,3], threshold=0) as a supposedly fresh call returns [100, 8, -9, 1, 2] — it's still holding state from the previous call. This is a classic Python trap, comes up constantly in interviews, and the fix is:
Never use a mutable object ([], {}, set()) as a default argument — use None and initialize inside.

````python
def process(samples, threshold = 5,results = []):

    for i in range(0, len(samples)):
        if samples[i] > threshold or samples[i] < -threshold:
            results.append(samples[i])
    return results

data = [1, 8, -9, 3, 6]
data[0] = 100

try:
    out = process(data, threshold=5)
    print(out)
except IndexError:
    print("bad index")
except:
    print('something wrong')
process([1,2,3], threshold=0)

    
````

````python
def analyze(signal, window) :
    windowsum = 0
    moving_average = []
    peak_indices = []
    if len(signal) < window:
        raise ValueError(f"window ({window}) larger than signal length ({len(signal)})")
    for i in range(window):
        windowsum += signal[i]
    moving_average.append(windowsum/window)
    start = 0
    end = window
    while end < len(signal) :
        windowsum  = windowsum - signal[start] + signal[end]
        start += 1
        end += 1
        moving_average.append(windowsum/window)
    j = 1
    while j < len(signal) -1 :
        if signal[j] > signal[j-1] and signal[j] > signal[j+1]:
            peak_indices.append(j)
        j += 1
   

    return moving_average, peak_indices
signal = [2, 4, 1, 5, 9, 3, 7, 2, 8, 6]
a, b = analyze(signal, window = 3)
print(f"Moving average : {a} and Peak Indices : {b}")
    
    

    
        
````

````python
from math import sin, cos, pi
def generate_signal(freq, sample_rate, duration, waveform="sine"):
    ans = []
    x = 0
    count = duration * sample_rate
    interval = 1/sample_rate
    c = 0
    while c  < count :
       if waveform == "sine":
         ans.append( sin(x*2 * pi * freq))
       elif waveform == 'square':
         if (sin(x*2 * pi * freq) >= 0) :
            ans.append(1)
         else :
            ans.append(-1)
       else :
         raise(ValueError)
       x += interval
       c +=1
    return ans
sinew = generate_signal(5,50,0.2,"sine")
for x in sinew[:5] :
   print(f"{x:0.4f}")
sqew = generate_signal(5,50,0.2,"square")
for x in sqew[:5] :
   print(f"{x:0.4f}")
          
          
````

````python
def detect_and_fix(signal, threshold):
    dict = {"pos" :0, "neg": 0, "zero" : 0}
    if len(signal) == 0:
        raise(ValueError("you are cooked"))
    for x in signal :
       if x > 0 :
           dict["pos"] += 1
       elif x < 0 :
           dict["neg"] += 1
       else :
           dict["zero"] += 1

    clipped = []
    for x in signal :
        if x > threshold:
            clipped.append(threshold)
        elif x < -threshold:
            clipped.append(-threshold)
        else :
            clipped.append(x)
    return  dict,clipped
    
signal = [3, -8, 0, 5, -2, 10, -10, 1]
print(detect_and_fix(signal,6))
````

````python
def time_reverse(x: np.ndarray) -> np.ndarray:
    return x[::-1]

def even_odd_decompose(x: np.ndarray) -> tuple[np.ndarray, np.ndarray]:
    xe = (x + time_reverse(x)) / 2
    xo = (x - time_reverse(x)) / 2
    return xe, xo
````

````python
import numpy as np

t_original = np.array([0.0, 2.0, 4.0, 6.0, 8.0])
x_original = np.array([0, 10, 20, 30, 40])
dt = 2.0

t_query = np.array([1.0, 5.0, 7.0])

idx_float = (t_query - t_original[0]) / dt
print(idx_float)

i_left = np.floor(idx_float).astype(int)
i_right = i_left + 1
print(i_left, i_right)

result = (x_original[i_left] + x_original[i_right]) / 2
print(result)
````
