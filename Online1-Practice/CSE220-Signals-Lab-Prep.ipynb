```python
fs = 100 # samples per second (sampling rate)
t = np.arange(-5, 5, 1/fs) # time axis
x = np.exp(-np.abs(t)) * np.cos(2*np.pi*t) # example base signal (given in template)
plt.plot(t, x, label="x(t)", color="blue")
plt.title("Signal x(t)")
plt.xlabel("t")
plt.ylabel("Amplitude")
plt.legend()
plt.grid(True)
plt.show()
```

```python
def time_reverse(signal):
    """Returns the time-reversed version of a sampled signal."""
    return signal[::-1] # reverse the array order

x_rev = time_reverse(x) # this represents x(-t) IF t is symmetric about 0
x_neg = time_reverse(x) # x(-t), using the required function
x_even = 0.5 * (x + x_neg) # xe(t)
x_odd = 0.5 * (x - x_neg) # xo(t)
# Sanity check: x_even should look symmetric, x_odd anti-symmetric about t=0
# Also: x_even + x_odd should reconstruct x exactly
print(np.allclose(x_even + x_odd, x)) # should print True
```

```python
def interpolate_signal(x, idx):
    """
    Returns an interpolated value of x at a (possibly fractional) index idx,
    using the average of the nearest left and right known samples.
    If idx lands exactly on an integer sample, return that sample directly.
    """
    left = int(np.floor(idx))
    right = int(np.ceil(idx))
    if left == right: # idx is already an integer index
        return x[left]
    if left < 0 or right >= len(x): # out of range -> caller should discard
        return None
    return 0.5 * (x[left] + x[right])
def time_scale(x, k):
    """Returns y(t) = x(t/k) for positive integer k, using interpolation for
    the newly required in-between samples. Discards points beyond range."""
    n = len(x)
    y = []
    for i in range(n):
        idx = i / k # fractional index into original x
        val = interpolate_signal(x, idx)
        if val is not None:
            y.append(val)
    return np.array(y)
x_demo = np.array([0, 4, 8, 12, 16]) # toy example, x(n) = 4n
y_demo = time_scale(x_demo, 2)
print(y_demo) # [0. 2. 4. 6. 8. 10. 12. 14. 16.] (length shrinks near the tail
# because higher indices go out of range and are discarded)
```

```python
def sinusoid(n, A, Omega0, phi):
    return A * np.cos(Omega0 * n + phi)
n = np.arange(0, 50, 1)
A, Omega0, phi = 2, 0.3, 0
x_n = sinusoid(n, A, Omega0, phi)
```

```python
def time_shift_sinusoid(n, A, Omega0, phi, n0):
    """Shifts the sinusoid in time by n0 samples: x[n - n0]."""
    return A * np.cos(Omega0 * (n - n0) + phi)
def phase_change_sinusoid(n, A, Omega0, phi, delta_phi):
    """Changes only the phase term, time axis n is untouched."""
    return A * np.cos(Omega0 * n + phi + delta_phi)


n0 = 5
delta_phi_equiv = -Omega0 * n0 # the matching phase change
x_shifted = time_shift_sinusoid(n, A, Omega0, phi, n0)
x_phase = phase_change_sinusoid(n, A, Omega0, phi, delta_phi_equiv)
mse = np.mean((x_shifted - x_phase)**2)
print("MSE:", mse) # should be ~0 when delta_phi matches -Omega0*n0
# Compare against a MISMATCHED phase change to show they normally differ:
x_phase_wrong = phase_change_sinusoid(n, A, Omega0, phi, delta_phi=0.5)
mse_wrong = np.mean((x_shifted - x_phase_wrong)**2)
print("MSE (mismatched):", mse_wrong) # clearly larger than ~0
```

```python
n = int(input())
events = []
for _ in range(n):
    label, amp, dur = input().split()
    amp, dur = int(amp), int(dur)
    events.append((label, amp, dur))
    
summary = {}
for label, amp, dur in events:
    if label not in summary:
        summary[label] = {"total_dur": 0, "total_energy": 0}
    summary[label]["total_dur"] += dur
    summary[label]["total_energy"] += amp * amp * dur

for label in sorted(summary.keys()):
    s = summary[label]
    print(label, s["total_dur"], s["total_energy"])

top_label = sorted(summary.keys(),
            key=lambda L: (-summary[L]["total_energy"], L))[0]
print("TOP", top_label, summary[top_label]["total_energy"])
```
