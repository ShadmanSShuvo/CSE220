```python
import numpy as np

# GIVEN: index array and value array. Nothing else. Zero everywhere outside.
n = np.array([-1, 0, 1, 2, 3])
x = np.array([ 4, 7, 2, 9, 5])

def x_at(query_idx, n_samp, x_samp):
    """
    Look up x[query_idx] for a discrete signal.
    Returns 0 wherever the index is not in n_samp (including non-integers).
    query_idx may be an array.
    """
    q = np.asarray(query_idx)
    out = np.zeros(q.shape, dtype=float)
    # match each queried index against the known index set
    for i, qi in np.ndenumerate(q):
        hit = np.where(n_samp == qi)[0]        # exact integer match only
        if hit.size:                           # found it -> copy value
            out[i] = x_samp[hit[0]]
        # else stays 0  (index unknown OR non-integer)
    return out

# sanity: identity
print("x[n] itself :", x_at(n, n, x))

# x[-n]  (reversal): query at -n over a chosen output range
n_out = np.arange(-3, 2)
print("n_out       :", n_out)
print("x[-n]       :", x_at(-n_out, n, x))   # note: argument is -n_out

```

```python
import numpy as np

n = np.array([-1, 0, 1, 2, 3])
x = np.array([ 4, 7, 2, 9, 5], dtype=float)

# Query a mix: some in range, one out of range, one non-integer
query = np.array([3, 2, 1.5, 0, 10])
print("query        :", query)

# STEP 1: which queries are integers?
is_int = (query == np.round(query))
print("np.round     :", np.round(query))
print("is_int       :", is_int, "  <- 1.5 is False")

# STEP 2: round to int form (so we can compare against integer index set)
qi = np.round(query).astype(int)
print("qi           :", qi)

# STEP 3: searchsorted -> where each qi WOULD be inserted to keep n sorted
pos = np.searchsorted(n, qi)
print("n            :", n)
print("pos (raw)    :", pos, "  <- note: 10 gives pos=5, off the end!")

# STEP 4: clip so pos is always a legal index into n (avoids crash on next line)
pos = np.clip(pos, 0, len(n) - 1)
print("pos (clipped):", pos)

# STEP 5: verify each slot ACTUALLY matches (searchsorted alone doesn't confirm a hit)
print("n[pos]       :", n[pos])
print("n[pos]==qi   :", (n[pos] == qi))
valid = is_int & (n[pos] == qi)
print("valid        :", valid, "  <- must be integer AND real match")

# STEP 6: write only the valid ones; everything else stays 0
out = np.zeros(query.shape, dtype=float)
out[valid] = x[pos[valid]]
print("out          :", out)
```
