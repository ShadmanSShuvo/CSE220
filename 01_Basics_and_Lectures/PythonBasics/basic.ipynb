```python
def iseven(num1, num2 = 9) -> int :
    "a function to check the similarity in parity of two numbers"
    k = 0
    if (num1 %2 == 0 and num2 % 2 == 0) or (num1 % 2 != 0 and num2 % 2 != 0) :
        print('Same parity')
        k = 1
    else :
        print('Different parity')
        k = 2
    return k
    
print(iseven(4,5))
print(iseven(num2 = 7, num1 = 2))
print(iseven(7))
```

```python
def paritycheck(*numbs) -> str: #this str is just for show
    "a function to check the similarity in parity of two numbers"
    if(len(numbs) == 0) :
        print('No number given')
    else :
       k = 0
       if(numbs[0] % 2 == 0) :
           k = 1
       for x in numbs :
           t = k ^ (x % 2 == 0)
           if t != 0 :
               return 'Not same', 1
       return 'same', 0
print(paritycheck(4,8,6))
    
```
