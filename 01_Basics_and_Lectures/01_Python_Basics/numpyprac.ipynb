```python
import numpy as np
nparr = np.array([1,2,3,4,5,6])
nparr2 = nparr.reshape((2,3))
print(nparr)
print(nparr2)
nparr2[1][2] = 9
print(nparr)
print(nparr2)
#just view ta change kore, tai jekono ta modify korle both reflect korbe
```

```python
nparr3 = nparr2.transpose()
print(nparr2)
print(nparr3)
nparr3[2][1] = 10
print(nparr2)
print(nparr3)
#changes in both places
```

```python
nparr4 = nparr[1:4]
nparr4[0] = 11
nparr
#splicing also is just a view
```

```python
nparr5 = np.sort(nparr)
print(nparr5)
print(nparr)
#two different arrays
```

```python
nparr6 = [3,4,99, 100]
nparr7 = np.concatenate((nparr, nparr6)) #etao alada copy toiri kore, not view
nparr7
```

```python
nparr7[0] = 998
nparr
```

```python
a = np.array([11,11,12,13,14,15,16,16,17,17,18,19,19,50,14,15])
a
```

```python
freqofara
```

```python
result = np.concatenate((unara,freqofara))
result = result.reshape(2,10)
result = result.transpose()
result
```
