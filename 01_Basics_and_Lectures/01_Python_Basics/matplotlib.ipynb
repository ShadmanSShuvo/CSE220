```python
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
plt.style.use('ggplot')
```

```python
t = np.linspace(0,10,100)
x = np.sin(t)

plt.Figure(figsize = (10,4))
plt.plot(t,x)
plt.title("5Hz System Input")        # The Title [cite: 1360]
plt.xlabel("Time (seconds)")         # X-Axis Label [cite: 1386]
plt.ylabel("Amplitude (Volts)")      # Y-Axis Label [cite: 1368]
plt.grid(True)                       # Adds a grid for easy reading [cite: 1363]
plt.legend(loc="upper right")        # Displays the label we defined in plt.plot() [cite: 1365]
plt.show
```

```python
voltage = np.linspace(0,10,100)

voltage_out = 0.5 * np.sin(2 * np.pi * 5 * t - (np.pi/4))

plt.figure(figsize=(10, 4))
plt.plot(t, voltage, label="Input $x(t)$")
plt.plot(t, voltage_out, label="Output $y(t)$", color="red")
plt.legend()
plt.show()
```

```python
fig, axs = plt.subplots(2, 1, figsize=(10, 6), layout="constrained") #(row = 2, column = 1 here)

# Plot on the Top Graph (Index 0)
axs[0].plot(t, voltage,label = 'blue sig', color="blue", )
axs[0].set_title("Input Signal")
axs[0].grid(True)
axs[0].legend()
#axs[row, col]
# Plot on the Bottom Graph (Index 1)
axs[1].plot(t, voltage_out, color="red")
axs[1].set_title("Attenuated Output")
axs[1].set_xlabel("Time (s)")
axs[1].grid(True)

# Render the whole figure
#plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# A discrete time axis (integers only)
n = np.array([-2, -1, 0, 1, 2, 3])
# The discrete signal values
x_n = np.array([0.0, 0.0, 1.0, 2.5, 1.0, -0.5])

plt.figure(figsize=(8, 3))

# Using stem instead of plot
# linefmt: changes the vertical stem lines
# markerfmt: changes the dot at the top
# basefmt: changes the x-axis baseline appearance
plt.stem(n, x_n, linefmt='b-', markerfmt='bo', basefmt='k-')

plt.title("Discrete Signal $x[n]$")
plt.xlabel("Sample Index (n)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```

```python
# The original continuous-looking signal
t_original = np.linspace(0, 2, 100)
x_original = np.sin(2 * np.pi * t_original)

# Your newly interpolated sample points
t_new = np.array([0.25, 0.75, 1.25, 1.75])
x_interpolated = np.sin(2 * np.pi * t_new) # Assuming perfect interpolation

plt.figure(figsize=(8, 3))

# 1. Plot the continuous wave first
plt.plot(t_original, x_original, label="Original $x(t)$", color='blue')

# 2. Overlay the specific points using scatter
# s=50 controls the size of the dots
# zorder=5 forces the dots to render ON TOP of the line
plt.scatter(t_new, x_interpolated, color='red', s=50, zorder=5, label="Interpolated Points")

plt.title("Interpolation Verification")
plt.legend()
plt.grid(True)
plt.show()
```

```python
t = np.linspace(-2, 2, 500)
# A parabolic function like the one in your FT assignment
y = t**2 

plt.figure(figsize=(8, 3))

# Plot the outer boundary line
plt.plot(t, y, color='black', linewidth=2)

# Shade the area under the curve
# alpha=0.3 makes the fill slightly transparent so you can see grid lines through it
plt.fill_between(t, y, 0, color='orange', alpha=0.3, label="Area = Integral")

plt.title("Numerical Integration bounds")
plt.legend()
plt.grid(True)
plt.show()
```
