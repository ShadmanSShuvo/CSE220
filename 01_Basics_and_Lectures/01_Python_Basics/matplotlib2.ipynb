Good instinct — matplotlib has a bunch of these shorthand-to-fullname mappings. Here are the main ones worth knowing:

## Linestyles
| String | Name |
|---|---|
| `'-'` | solid |
| `'--'` | dashed |
| `'-.'` | dashdot |
| `':'` | dotted |
| `'None'`, `' '`, `''` | none (no line) |

## Markers (most common)
| String | Name/Shape |
|---|---|
| `'.'` | point |
| `','` | pixel |
| `'o'` | circle |
| `'v'` | triangle_down |
| `'^'` | triangle_up |
| `'<'` | triangle_left |
| `'>'` | triangle_right |
| `'s'` | square |
| `'p'` | pentagon |
| `'*'` | star |
| `'h'` | hexagon1 |
| `'H'` | hexagon2 |
| `'+'` | plus |
| `'x'` | x |
| `'D'` | diamond |
| `'d'` | thin_diamond |
| `'|'` | vline |
| `'_'` | hline |

## Colors (single-letter shorthand)
| String | Name |
|---|---|
| `'b'` | blue |
| `'g'` | green |
| `'r'` | red |
| `'c'` | cyan |
| `'m'` | magenta |
| `'y'` | yellow |
| `'k'` | black |
| `'w'` | white |

## Drawstyles
| String | Description |
|---|---|
| `'default'` | straight line between points |
| `'steps'` / `'steps-pre'` | step before each point |
| `'steps-mid'` | step at midpoint |
| `'steps-post'` | step after each point |

## Other common kwarg aliases
| Short | Long |
|---|---|
| `c` | `color` |
| `lw` | `linewidth` |
| `ls` | `linestyle` |
| `ms` | `markersize` |
| `mfc` | `markerfacecolor` |
| `mec` | `markeredgecolor` |
| `mew` | `markeredgewidth` |

Great catch — this is the classic **pyplot (state-based) vs object-oriented (OO) API** distinction. When you work with `ax` (from `subplots()` or `axes()`), most `plt.` functions become `ax.` methods, and labeling functions get a `set_` prefix. Here's the full mapping:

## Labeling — gets `set_` prefix
| `plt.` (pyplot style) | `ax.` (OO style) |
|---|---|
| `plt.title('...')` | `ax.set_title('...')` |
| `plt.xlabel('...')` | `ax.set_xlabel('...')` |
| `plt.ylabel('...')` | `ax.set_ylabel('...')` |
| `plt.xlim(min, max)` | `ax.set_xlim(min, max)` |
| `plt.ylim(min, max)` | `ax.set_ylim(min, max)` |
| `plt.xticks([...])` | `ax.set_xticks([...])` |
| `plt.yticks([...])` | `ax.set_yticks([...])` |

## Same name, no prefix change
| `plt.` | `ax.` |
|---|---|
| `plt.plot(x, y)` | `ax.plot(x, y)` |
| `plt.scatter(x, y)` | `ax.scatter(x, y)` |
| `plt.bar(x, h)` | `ax.bar(x, h)` |
| `plt.hist(data)` | `ax.hist(data)` |
| `plt.legend()` | `ax.legend()` |
| `plt.grid(True)` | `ax.grid(True)` |
| `plt.annotate(...)` | `ax.annotate(...)` |
| `plt.text(x, y, s)` | `ax.text(x, y, s)` |
| `plt.fill_between(...)` | `ax.fill_between(...)` |

## Special renamed cases
| `plt.` | `ax.` equivalent | Note |
|---|---|---|
| `plt.axis('equal')` | `ax.set_aspect('equal')` | different name entirely |
| `plt.axis([xmin,xmax,ymin,ymax])` | `ax.axis([xmin,xmax,ymin,ymax])` | actually stays the same here |
| `plt.tick_params(...)` | `ax.tick_params(...)` | same |
| `plt.colorbar()` | `fig.colorbar(mappable, ax=ax)` | needs `fig`, and the mappable object |

## Things that stay on `plt`/`fig`, not `ax`
| Function | Why |
|---|---|
| `plt.show()` | displays the whole figure, not one axes |
| `plt.savefig(...)` | saves the whole figure |
| `fig.suptitle('...')` | title for the *whole figure* (multiple subplots), vs `ax.set_title()` for one subplot |
| `plt.tight_layout()` | adjusts whole figure layout (though `fig.tight_layout()` also works) |



**Rule of thumb for exams:** if the function *sets a property* (title, label, limits, ticks) → add `set_` on `ax`. If the function *draws something* (plot, scatter, bar, legend, grid, annotate) → same name, just called on `ax` instead of `plt`. Only `show()`, `savefig()`, and figure-wide titles (`suptitle`) stay separate from `ax`.

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [1, 4, 9, 16]

# ---- pyplot style ----
plt.plot(x, y, color='blue', label = 'xdd')
plt.title("Pyplot Style")
plt.xlabel("x")
plt.ylabel("y")
plt.xlim(0, 5)
plt.legend()
plt.grid(True)
plt.show()

# ---- OO style ----
fig, ax = plt.subplots()
ax.plot(x, y, color='blue', label='y = x²')
ax.set_title("OO Style")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_xlim(0, 5)
ax.legend()
ax.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

categories = ['A', 'B', 'C', 'D']
values = [23, 45, 12, 36]

plt.bar(categories, values, width=0.6, color='skyblue', edgecolor='black')
plt.title("Vertical Bar Chart")
plt.xlabel("Category")
plt.ylabel("Value")
plt.show()
```

```python
plt.bar(categories, values, yerr=[2, 3, 1, 4], capsize=5, color='orange')
plt.show()
```

```python
values2 = [10, 15, 5, 20]
plt.bar(categories, values, color='skyblue', label='Group 1')
plt.bar(categories, values2, bottom=values, color='salmon', label='Group 2')
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [3, 7, 2, 8, 5]

plt.stem(x, y, linefmt='b-', markerfmt='ro', basefmt='k-')
plt.title("Stem Plot")
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y1 = [1, 3, 5, 7, 9]
y2 = [2, 2, 4, 5, 6]

plt.plot(x, y1, color='blue', label='y1')
plt.plot(x, y2, color='red', label='y2')
plt.fill_between(x, y1, y2, color='gray', alpha=0.3)
plt.legend()
plt.title("Fill Between Two Curves")
plt.show()
```
